```python
# 6. Conditionals 

Employee_Name = input("Enter your name: ")
KPI_Score = int(input("Enter your KPI Score: "))

if KPI_Score < 0 or KPI_Score > 100:
    print("Invalid score")

elif KPI_Score >= 90:
    status = "Exceptional"

elif KPI_Score >= 75:
    status = "Commendable"

elif KPI_Score >= 60:
    status = "Satisfactory"

else:
    status = "Needs Review"

if 0 <= KPI_Score <= 100:
    print(f"Employee: {Employee_Name} Score: {KPI_Score} Status: {status}")
else:
    print("Invalid score")
```
